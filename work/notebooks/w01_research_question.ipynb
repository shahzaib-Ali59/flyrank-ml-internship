## 2. The question: decision, action, cost of a wrong call

**Question:** Which pages should a content reviewer look at first when deciding whether to
refresh, expand, protect, or leave a page alone?

**Unit of analysis:** one page (one row = one content item, using `content_id` as the grain).

**Decision this improves:** which pages a content team spends its limited review time on
this cycle, instead of reviewing pages in arbitrary order or relying only on gut feel.

**Who acts on it:** a content strategist or SEO reviewer with a fixed weekly capacity — say,
they can only manually review 20-50 pages a week.

**Cost of a wrong recommendation:**
- False positive (flagging a healthy page as needing review): wastes reviewer time that
  could've gone to a page that actually needed it.
- False negative (missing a genuinely declining page): the page keeps losing visibility and
  traffic unnoticed until the next review cycle, which could mean weeks of lost clicks.

**Why data/ML can help at all:** with tens of thousands of pages and limited reviewer
capacity, someone has to rank them. A hand-written rule (like "flag anything not updated in
180 days") is fast but crude — my Week 1 numbers already showed a learned model can rank far
more precisely (Precision@50 of 0.740 vs. 0.240), meaning reviewers spend their time on pages
that are actually worth it far more often.

## 3. Quick look at the data (2-3 real numbers)

Loading the starter dataset to back up why this lane is worth pursuing.

```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

total_pages = len(df)
declining_pages = (df["trend_direction"].str.lower() == "down").sum()
declining_pct = declining_pages / total_pages * 100

stale_visible = ((df["days_since_last_update"] >= 180) & (df["impressions_90d"] >= 500)).sum()
stale_visible_pct = stale_visible / total_pages * 100

print(f"Total pages in starter dataset: {total_pages:,}")
print(f"Pages currently declining: {declining_pages:,} ({declining_pct:.1f}%)")
print(f"Stale + visible pages (>=180 days since update, >=500 impressions): "
      f"{stale_visible:,} ({stale_visible_pct:.1f}%)")
print(f"\nFrom Week 1: a hand-written rule scored Precision@50 = 0.240, "
      f"a random forest scored 0.740 -- a ~3x improvement in ranking quality.")
```

## 4. Careful words: what I can and can't claim

**What I can say:** which pages show observed signals of decline, low visibility, or
low engagement relative to similar pages, and rank them by review priority using patterns
learned from historical data. This is decision-support: it helps a human prioritize, not
replace their judgment.

**What I cannot say:** that a page will definitely recover if refreshed (that requires a
causal experiment, not this data), that I've reverse-engineered Google's ranking algorithm,
or that a "declining" label proves real decline rather than seasonality, consolidation with
a sibling page, or noise -- all of which I'll need to rule out as I go deeper.

All findings will be described as observed and directional, never as proof.
